```python
import gradio as gr
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM

# -------------------------------
# Load English → Urdu model
# -------------------------------
tokenizer_en_ur = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-ur")
model_en_ur = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-ur")

en_to_ur = pipeline(
    "translation_en_to_ur",
    model=model_en_ur,
    tokenizer=tokenizer_en_ur
)

# -------------------------------
# Load Urdu → English model
# -------------------------------
tokenizer_ur_en = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-ur-en")
model_ur_en = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-ur-en")

ur_to_en = pipeline(
    "translation_ur_to_en",
    model=model_ur_en,
    tokenizer=tokenizer_ur_en
)

# -------------------------------
# Translation Logic
# -------------------------------
def translate_text(text, direction):
    if not text.strip():
        return "Please enter some text to translate."

    try:
        if direction == "English → Urdu":
            result = en_to_ur(text, max_length=512)
        else:
            result = ur_to_en(text, max_length=512)

        return result[0]["translation_text"]

    except Exception as e:
        return f"Error: {str(e)}"

# -------------------------------
# Swap Function
# -------------------------------
def swap_languages(direction):
    if direction == "English → Urdu":
        return "Urdu → English"
    return "English → Urdu"

# -------------------------------
# Gradio UI
# -------------------------------
with gr.Blocks(title="English ↔ Urdu Translator") as app:

    gr.Markdown("## English ↔ Urdu Translator")
    gr.Markdown("Simple translation using Hugging Face models.")

    with gr.Row():
        direction = gr.Dropdown(
            ["English → Urdu", "Urdu → English"],
            value="English → Urdu",
            label="Translation Direction"
        )

        swap_btn = gr.Button("Swap")

    input_text = gr.Textbox(
        label="Enter Text",
        placeholder="Type here...",
        lines=4
    )

    translate_btn = gr.Button("Translate")

    output_text = gr.Textbox(
        label="Translated Text",
        lines=4
    )

    gr.Examples(
        examples=[
            ["Hello, how are you?", "English → Urdu"],
            ["آپ کیسے ہیں؟", "Urdu → English"]
        ],
        inputs=[input_text, direction],
    )

    translate_btn.click(
        fn=translate_text,
        inputs=[input_text, direction],
        outputs=output_text
    )

    swap_btn.click(
        fn=swap_languages,
        inputs=direction,
        outputs=direction
    )

# Run app
if __name__ == "__main__":
    app.launch()
```
